# Объяснение датасета

Пропуски есть только в dataframe'е с книгами в признаках: isbn, isbn13, original_publication_year, original_title, language_code

Главная неочевидная особенность - у книг может быть несколько изданий. Несколько названий(локализация и оригинал).
И несколько id - сервиса и стандартизованные библиографичные. В датафрейме с книгами уже есть агрегированные данные по оценкам, которые можно было самостоятельно получить по датафрейму с оценками.

1) Датафрейм с книгами - разберем подробнее ниже
2) Датафрейм с оценками - id издания книги, пользователь, оценка
3) Датафрейм с тегами книг - id издания книги, тег, количество(сколько раз юзеры протегировали)
4) Датафрейм с тегами - id, имя
5) Датафрейм с прочитанными книгами - пользователь, id издания книги

Пропуски в данных есть только в книгах. Подробнее о его фичах(получено с помощью LLM):

### **`book_id`**
- **Тип**: int64
- **Описание**: Внутренний уникальный идентификатор записи в датасете
- **Пример**: 1, 2, 3, ...

### **`goodreads_book_id`**
- **Тип**: int64  
- **Описание**: Уникальный идентификатор книги в системе GoodReads
- **Важность**: Ключ для соединения с ratings_df
- **Пример**: 2767052, 3, 41865

### **`best_book_id`**
- **Тип**: int64
- **Описание**: Идентификатор "лучшего" издания книги (может отличаться от book_id если есть разные издания)
- **Пример**: 2767052, 3, 41865

### **`work_id`**
- **Тип**: int64
- **Описание**: Идентификатор литературного произведения (work), которое может иметь несколько изданий
- **Пример**: 2792775, 4640799

## 📖 Библиографическая информация

### **`books_count`**
- **Тип**: int64
- **Описание**: Количество различных изданий этой книги в системе
- **Интерпретация**: Высокое значение = популярное произведение с множеством изданий
- **Пример**: 50, 120, 5

### **`isbn`**
- **Тип**: object (строка)
- **Пропуски**: 700 (7%)
- **Описание**: International Standard Book Number (10-значный)
- **Пример**: "439554934", "743477111"

### **`isbn13`**
- **Тип**: float64
- **Пропуски**: 585 (5.85%)
- **Описание**: 13-значный ISBN (современный стандарт)
- **Пример**: 9780439554930.0, 9780743477116.0

### **`authors`**
- **Тип**: object (строка)
- **Описание**: Автор(ы) книги. Если несколько авторов - разделены запятыми
- **Пример**: "J.K. Rowling", "Stephen King, Peter Straub"

### **`original_publication_year`**
- **Тип**: float64
- **Пропуски**: 21 (0.21%)
- **Описание**: Год первого издания книги
- **Пример**: 1997.0, 2003.0, 1813.0

### **`original_title`**
- **Тип**: object (строка)
- **Пропуски**: 585 (5.85%)
- **Описание**: Оригинальное название на языке публикации
- **Пример**: "Harry Potter and the Philosopher's Stone"

### **`title`**
- **Тип**: object (строка)
- **Описание**: Название книги (может быть локализованным)
- **Пример**: "Harry Potter and the Sorcerer's Stone" (американская версия)

### **`language_code`**
- **Тип**: object (строка)
- **Пропуски**: 1084 (10.84%)
- **Описание**: Код языка книги
- **Пример**: "eng", "spa", "fre", "rus"

## ⭐ Рейтинги и отзывы

### **`average_rating`**
- **Тип**: float64
- **Описание**: Средний рейтинг книги по шкале 1-5
- **Важность**: Ключевой показатель качества
- **Пример**: 4.57, 3.82, 2.15

### **`ratings_count`**
- **Тип**: int64
- **Описание**: Общее количество оценок этой книги
- **Важность**: Показатель популярности

### **`work_ratings_count`**

### **`work_text_reviews_count`**

```python
user_activity = ratings_df.groupby('user_id').agg(
    rating_count=('rating', 'count')
).reset_index()
user_activity = user_activity.sort_values('rating_count', ascending=False)
```

```python
mu = user_activity['rating_count'].mean()
sigma = user_activity['rating_count'].std()

plt.figure(figsize=(16, 12))

n, bins, patches = plt.hist(
    user_activity['rating_count'],
    bins=50,
    alpha=0.7,
    color='skyblue',
    edgecolor='black',
)

xmin, xmax = plt.xlim()
x = np.linspace(xmin, xmax, 100)
p = stats.norm.pdf(x, mu, sigma) * len(user_activity) * (bins[1] - bins[0])
plt.plot(x, p, 'r-', linewidth=2, label=f'Нормальное распределение\n(μ={mu:.1f}, σ={sigma:.1f})')

plt.title('Взаимоотношение количества пользователей и количества оценок')
plt.xlabel('Количество оценок')
plt.ylabel('Количество пользователей')
plt.grid(True, alpha=0.3)
plt.legend()

stats_text = f'''
Статистики распределения:
Общее количество пользователей: {len(user_activity):,}
Общее количество оценок: {user_activity['rating_count'].sum():,}
Среднее: {mu:.1f} оценок/пользователя
Медиана: {user_activity['rating_count'].median():.1f}
Стандартное отклонение: {sigma:.1f}
Максимум: {user_activity['rating_count'].max():.1f}
Минимум: {user_activity['rating_count'].min():.1f}
'''
plt.annotate(
    stats_text,
    xy=(0.02, 0.98),
    xycoords='axes fraction',
    bbox=dict(boxstyle="round,pad=0.3", facecolor="white", alpha=0.8),
    verticalalignment='top',
    fontsize=10
)

plt.tight_layout()
plt.show()
```

```python
def check_user_activity_normality(user_activity):
    data = user_activity['rating_count']

    skewness = stats.skew(data)
    kurtosis = stats.kurtosis(data)

    if abs(skewness) < 0.5:
        skew_interpretation = "примерно симметричное"
    elif skewness > 0:
        skew_interpretation = "правосторонняя асимметрия"
    else:
        skew_interpretation = "левосторонняя асимметрия"

    if abs(kurtosis) < 0.5:
        kurt_interpretation = "нормальный эксцесс"
    elif kurtosis > 0:
        kurt_interpretation = "островершинное (тяжелые хвосты)"
    else:
        kurt_interpretation = "плосковершинное (легкие хвосты)"

    print(f"{skew_interpretation}, {kurt_interpretation}")
    print()

    # Статистические тесты на нормальность
    alpha = 0.05
    # Тест Д'Агостино (подходит для больших выборок)
    stat_da, p_da = normaltest(data)
    print("📈 Тест Д'Агостино на нормальность:")
    print(f"   Вывод: {'Распределение нормальное' if p_da > alpha else 'Распределение НЕ нормальное'}")
    print()

    # Анализ правила 68-95-99.7
    within_1std = ((data >= data.mean() - data.std()) & (data <= data.mean() + data.std())).mean() * 100
    within_2std = ((data >= data.mean() - 2*data.std()) & (data <= data.mean() + 2*data.std())).mean() * 100
    within_3std = ((data >= data.mean() - 3*data.std()) & (data <= data.mean() + 3*data.std())).mean() * 100

    print("📐 Правило 68-95-99.7 для нормального распределения:")
    print(f"   В пределах 1σ: {within_1std:.1f}% (ожидается 68.3%)")
    print(f"   В пределах 2σ: {within_2std:.1f}% (ожидается 95.4%)")
    print(f"   В пределах 3σ: {within_3std:.1f}% (ожидается 99.7%)")
    print()

    # Анализ перцентилей
    percentiles = np.percentile(data, [25, 50, 75, 90, 95, 99])
    print("🎯 Перцентили распределения:")
    print(f"   25-й перцентиль: {percentiles[0]:.1f} оценок")
    print(f"   50-й перцентиль: {percentiles[1]:.1f} оценок")
    print(f"   75-й перцентиль: {percentiles[2]:.1f} оценок")
    print(f"   90-й перцентиль: {percentiles[3]:.1f} оценок")
    print(f"   95-й перцентиль: {percentiles[4]:.1f} оценок")
    print(f"   99-й перцентиль: {percentiles[5]:.1f} оценок")

check_user_activity_normality(user_activity)
```

<b>Распределение примерно нормальное.</b>

## Анализ популярности книг

```python
book_popularity = ratings_df.groupby('book_id').agg(
    rating_count=('rating', 'count'),
    avg_rating=('rating', 'mean')
).reset_index()
book_popularity = book_popularity.sort_values('rating_count', ascending=False)

display(book_popularity['rating_count'].describe())
print(f"\nМедиана количества оценок: {book_popularity['rating_count'].median()}")
```

```python
total_books = len(book_popularity)
total_ratings = book_popularity['rating_count'].sum()

tail_threshold = book_popularity['rating_count'].quantile(0.25)
head_threshold = book_popularity['rating_count'].quantile(0.75)

popular_books = book_popularity[book_popularity['rating_count'] >= head_threshold]
not_popular_books = book_popularity[book_popularity['rating_count'] <= tail_threshold]
medium_books = book_popularity[(book_popularity['rating_count'] > tail_threshold) & (book_popularity['rating_count'] < head_threshold)]

popular_books_ratings = popular_books['rating_count'].sum()
not_popular_books_ratings = not_popular_books['rating_count'].sum()
medium_books_ratings = medium_books['rating_count'].sum()
assert (popular_books_ratings + not_popular_books_ratings + medium_books_ratings) == total_ratings

print(f"Популярные книги {len(popular_books)/total_books*100:.1f}% создают {popular_books_ratings:,} оценок из {total_ratings:,}")
print(f"Это {popular_books_ratings/total_ratings*100:.1f}% всех оценок")

print(f"Непопулярные книги {len(not_popular_books)/total_books*100:.1f}% создают {not_popular_books_ratings:,} оценок из {total_ratings:,}")
print(f"Это {not_popular_books_ratings/total_ratings*100:.1f}% всех оценок")

print(f"Средние книги {len(medium_books)/total_books*100:.1f}% создают {medium_books_ratings:,} оценок из {total_ratings:,}")
print(f"Это {medium_books_ratings/total_ratings*100:.1f}% всех оценок")
```

```python
def get_title(row):
    orig_title = str(row['original_title'])
    eng_title = str(row['title'])

    if any('\u0590' <= char <= '\u06FF' for char in orig_title):
        return eng_title + ' [араб.]'
    else:
        return orig_title

book_popularity_merged = book_popularity.merge(
    books_df[['book_id', 'title', 'original_title']],
    on='book_id',
    how='left'
)
book_popularity_merged = book_popularity_merged[book_popularity_merged['original_title'].notna() & book_popularity_merged['title'].notna()]
book_popularity_merged = book_popularity_merged.sort_values('rating_count', ascending=False)

top_10_books = book_popularity_merged.head(10)
print("\nТОП-10 САМЫХ ПОПУЛЯРНЫХ КНИГ:\n")
for i, (_, row) in enumerate(top_10_books.iterrows(), 1):
    print(f"{i:2d}. {get_title(row):60} | {row['rating_count']:5d} оценок | {row['avg_rating']:.2f} ★")
```

```python
def create_book_profiles(books_df, book_tags_df, tags_df):
    books_with_tags = books_df[['book_id', 'goodreads_book_id', 'title', 'original_title']].merge(
        book_tags_df[['goodreads_book_id', 'tag_id', 'count']],
        on='goodreads_book_id',
        how='left'
    ).merge(
        tags_df[['tag_id', 'tag_name']],
        on='tag_id',
        how='left'
    )

    profiles = {}
    for _, row in books_with_tags.iterrows():
        book_id = row['book_id']
        # это итерация не просто по книгам, а по тегам, поэтому книга уже могла быть добавлена
        if book_id not in profiles:
            title = row['title']
            profiles[book_id] = {
                'title': title,
                'weighted_tags': [],
                'all_tags': set()
            }

        tag_name = str(row['tag_name']).lower().replace(' ', '_')
        count = max(int(row['count']), 1)
        # с помощью логарифмирования сокращаем объем профилей с миллионов до десятков тысяч
        log_weight = int(np.log2(count))
        profiles[book_id]['weighted_tags'].extend([tag_name] * log_weight)
        profiles[book_id]['all_tags'].add(row['tag_name'])

    book_profiles = []
    for book_id, data in profiles.items():
        book_profiles.append({
            'book_id': book_id,
            'title': data['title'],
            # будем использовать эту длинную строку для tfidf
            'profile': f"{data['title']} {' '.join(data['weighted_tags'])}".strip(),
            'tags_count': len(data['all_tags']),
            'all_tags': ' '.join(data['all_tags'])
        })

    return pd.DataFrame(book_profiles)
```

```python
def get_similar_books(book_id, book_profiles_df, tfidf_matrix, N=5):
    book_index = book_profiles_df[book_profiles_df['book_id'] == book_id].index[0]
    target_vector = tfidf_matrix[book_index:book_index + 1]
    cosine_similarities = cosine_similarity(
        target_vector,
        tfidf_matrix
    ).flatten()
    similar_indexes_asc = cosine_similarities.argsort()
    similar_indexes_desc = similar_indexes_asc[::-1]
    # [1:] чтобы исключить book_id книгу(нулевой индекс сама книга book_id)
    similar_indexes = similar_indexes_desc[1:N+1]

    similar_books = []
    for id in similar_indexes:
        similarity_score = cosine_similarities[id]
        similar_book_id = book_profiles_df.iloc[id]['book_id']
        similar_books.append({
            'book_id': similar_book_id,
            'similarity_score': similarity_score,
        })

    return pd.DataFrame(similar_books)
```

# Коллаборативная фильтрация

```python
def create_user_item_matrix(ratings_df):
    user_item_matrix = ratings_df.pivot_table(
        index='user_id',
        columns='book_id',
        values='rating',
        fill_value=0
    )

    return user_item_matrix
```

```python
user_item_matrix = create_user_item_matrix(ratings_df)

print("Матрица взаимодействий пользователь — книга:")
print(f"Размер: {user_item_matrix.shape}")
```

```python
def calculate_item_similarities(user_item_matrix):
    all_books = user_item_matrix.columns
    data = user_item_matrix.values.T
    # средняя оценка по всем книгам для каждого пользователя
    user_means = data.mean(axis=1, keepdims=True)
    # нормализуем оценки каждого пользователя(кто-то систематически завышает, кто-то занижает)
    normalized_data = data - user_means

    similarity_matrix = cosine_similarity(normalized_data)

    return similarity_matrix
```

```python
similarity_matrix = calculate_item_similarities(user_item_matrix)

print("Матрица попарных схожестей:")
print(f"Размер: {similarity_matrix.shape} (книги × книги)")
```

```python
def predict_rating_item_based(user_id, book_id, user_item_matrix, similarity_matrix, k=5):
    user_ratings = user_item_matrix.loc[user_id].values

    book_index = user_item_matrix.columns.get_loc(book_id)
    book_similarities = similarity_matrix[book_index]

    valid_indices = []
    valid_similarities = []
    for other_book_index, similarity in enumerate(book_similarities):
        rating = user_ratings[other_book_index]
        if similarity > 0 and rating > 0 and other_book_index != book_index:
            valid_indices.append(other_book_index)
            valid_similarities.append(similarity)
    if len(valid_similarities) == 0:
        return 0

    similarity_sum = 0
    weighted_sum = 0
    # лучшие схожести по книге book_id для user_id
    top_k_indices = np.argsort(valid_similarities)[::-1][:k]
    for i in top_k_indices:
        other_book_index = valid_indices[i]
        similarity = valid_similarities[i]

        rating = user_ratings[other_book_index]
        similarity_sum += similarity
        weighted_sum += similarity * rating
    if similarity_sum == 0:
        return 0

    predicted_rating = weighted_sum / similarity_sum
    return predicted_rating
```

```python
def get_item_based_recommendations(user_id, user_item_matrix, similarity_matrix, k=5):
    user_index = user_item_matrix.index.get_loc(user_id)
    user_ratings = user_item_matrix.iloc[user_index]
    unrated_books = user_ratings[user_ratings == 0].index

    # рекомендации по всем книгам, которые пользователь не оценивал(не видел)
    predictions = []
    for book_id in unrated_books:
        predicted_rating = predict_rating_item_based(
            user_id, book_id, user_item_matrix, similarity_matrix, k
        )
        if predicted_rating > 0:
            predictions.append({
                'book_id': book_id,
                'predicted_rating': predicted_rating,
            })
    predictions.sort(key=lambda x: x['predicted_rating'], reverse=True)
    return pd.DataFrame(predictions[:k])
```

Вычислительная сложность подхода:

1. **Расчет матрицы схожестей**: O(n² × m), где n - количество книг, m - количество пользователей
   - Попарное сравнение всех книг между собой
   - При 10,000 книг: ~100 млн операций

2. **Предсказание оценок**: O(n × k) на пользователя
   - Для каждого непросмотренного фильма проверяем k похожих
   - При 1,000 непросмотренных книг: 1,000 × k операций

3. **Память**: O(n²) для хранения матрицы схожестей
   - При 10,000 книг: 100 млн элементов (800 МБ)

Оптимизации:

### 1. **Семплирование и аппроксимация**
- Работать только с популярными книгами (top-N по количеству оценок)

### 2. **Разреженные представления**
- Хранить только значимые схожести (выше порога)
- Использовать разреженные матрицы и структуры данных§

### 3. **Индексирование и кэширование**
- Предварительно вычислять топ-K похожих для каждой книги
- Кэшировать результаты предсказаний

Будем оценивать рекомендации для каждого юзера для нескольких k. Чтобы посчитать метрики по методу возьмем N пользователей и усредним значения.

```python
def calculate_relevance_metrics(method, user_id, recommendations_book_ids, ratings, relevance_threshold=4):
    relevant_items = set(ratings[(ratings['user_id'] == user_id) & (ratings['rating'] >= relevance_threshold)]['book_id'].tolist())
    all_test_items = set(ratings[ratings['user_id'] == user_id]['book_id'].tolist())

    metrics = {
        'method': method,
    }

    for k in [3, 5, 10, 20, 30]:
        rec_k = recommendations_book_ids[:k]
        if len(rec_k) <= 0 or len(relevant_items) <= 0:
            metrics.update({
                f'precision@{k}': 0,
                f'recall@{k}': 0,
                f'ndcg@{k}': 0
            })
            continue

        relevant_in_rec = len(set(rec_k) & relevant_items)
        precision = relevant_in_rec / len(rec_k)

        recall = relevant_in_rec / len(relevant_items)

        dcg = 0
        for i, item in enumerate(rec_k):
            if item in relevant_items:
                dcg += 1 / np.log2(i + 2)
        ideal_relevance = min(len(relevant_items), k)
        idcg = sum(1 / np.log2(i + 2) for i in range(ideal_relevance))
        ndcg = dcg / idcg

        metrics.update({
            f'precision@{k}': precision,
            f'recall@{k}': recall,
            f'ndcg@{k}': ndcg
        })

    metrics.update({
        'relevant_items_count': len(relevant_items),
        'items_count': len(all_test_items)
    })

    return metrics
```

```python
# Признаки пользователей: средний рейтинг, количество оценок, активность.
class UserFeatures(BaseEstimator, TransformerMixin):
    def __init__(self):
        pass

    # X: ratings_df
    def fit(self, X, y=None):
        print('fit UserFeatures...', datetime.now().strftime('%H:%M'))
        self.user_mean_rating_ = X.groupby('user_id')['rating'].mean().rename('user_mean_rating')
        self.user_num_ratings_ = X.groupby('user_id')['rating'].count().rename('user_num_ratings')
        self.user_activity_ = X.groupby('user_id')['book_id'].nunique().rename('user_activity')
        return self

    def transform(self, X):
        print('transform UserFeatures...', datetime.now().strftime('%H:%M'))
        X = X.join(self.user_mean_rating_, on='user_id')
        X = X.join(self.user_num_ratings_, on='user_id')
        X = X.join(self.user_activity_, on='user_id')
        return X
```

```python
# Признаки книг: популярность, разнообразие оценок, тематические категории.
class BookFeatures(BaseEstimator, TransformerMixin):
    def __init__(self, books_df, book_tags_df, tags_df, top_n_tags=20):
        self.books_df = books_df
        self.book_tags_df = book_tags_df
        self.tags_df = tags_df
        self.top_n_tags = top_n_tags

    def fit(self, X, y=None):
        print('fit BookFeatures...', datetime.now().strftime('%H:%M'))
        self.book_popularity_ = X.groupby('book_id')['rating'].count().rename('book_popularity')
        self.book_rating_std_ = X.groupby('book_id')['rating'].std().fillna(0).rename('book_rating_std')

        # находим id топ-N тегов по встречаемости
        top_tags = (
            self.book_tags_df
            .groupby('tag_id')['count'].sum()
            .sort_values(ascending=False)
            .head(self.top_n_tags)
            .index
        )
        # фильтруем сведения о тегах книги, только если они в топе
        book_top_tags_df = (
            self.book_tags_df[self.book_tags_df['tag_id'].isin(top_tags)]
            .merge(
                self.books_df[['book_id', 'goodreads_book_id']],
                on='goodreads_book_id',
            )
        )
        book_tag_matrix = pd.crosstab(book_top_tags_df['book_id'], book_top_tags_df['tag_id'])

        # заменяет id тега на имя тега
        tag_id_to_name = dict(zip(self.tags_df['tag_id'], self.tags_df['tag_name']))
        book_tag_matrix.columns = [f'tag_{tag_id_to_name.get(tid, tid)}' for tid in book_tag_matrix.columns]

        self.book_tag_matrix_ = book_tag_matrix

        return self

    def transform(self, X):
        print('transform BookFeatures...', datetime.now().strftime('%H:%M'))
        X = X.join(self.book_popularity_, on='book_id')
        X = X.join(self.book_rating_std_, on='book_id')
        X = X.join(self.book_tag_matrix_, on='book_id')
        return X
```

```python
# Признаки взаимодействий: схожесть с историей пользователя.
class InteractionFeatures(BaseEstimator, TransformerMixin):
    def fit(self, X, y=None):
        print('fit InteractionFeatures...', datetime.now().strftime('%H:%M'))
        self.tag_columns = [col for col in X.columns if col.startswith('tag_')]
        return self

    def _user_sim(self, user_id, group):
        idxs = group.index.values
        tag_matrix = group[self.tag_columns].values
        sims = np.zeros(len(idxs))
        if tag_matrix.shape[0] == 1:
            return idxs, sims
        sim_matrix = cosine_similarity(tag_matrix)
        for i in range(len(idxs)):
            sim_sum = sim_matrix[i].sum() - 1.0
            mean_sim = sim_sum / (sim_matrix.shape[0] - 1)
            sims[i] = mean_sim
        return idxs, sims

    def transform(self, X):
        print('transform InteractionFeatures...', datetime.now().strftime('%H:%M'))

        sim_scores = np.zeros(X.shape[0])
        groups = list(X.groupby('user_id'))
        results = Parallel(n_jobs=-1)(
            delayed(self._user_sim)(user_id, group) for user_id, group in groups
        )
        for idxs, sims in results:
            sim_scores[idxs] = sims

        X['book_user_tag_similarity'] = sim_scores
        return X
```

## Создание моделей-scorer'ов: top-n, контетная, коллаборативная, matrix factorization

По возможности нужно группировать данные по пользователям, использовать разряженные структуры данных и кешировать промежуточные этапы тренировок, чтобы можно было быстрее переобучать pipeline. Также нужно предусмотреть, что использовать тренировочный кеш на тестовых данных нельзя, поэтому надо обязательно добавлять режим работы без кеша.
За основу брались мои модели из предыдущей работы, но их приходилось переделывать в целях оптимизации ресурсов. + они обернуты в классы для работы в pipeline.

```python
class ContentBasedScorer(BaseEstimator, TransformerMixin):
    def __init__(self, books_df, book_tags_df, tags_df,
                 max_features=15000, min_df=0.01, max_df=0.75):
        self.books_df = books_df
        self.book_tags_df = book_tags_df
        self.tags_df = tags_df
        self.max_features = max_features
        self.min_df = min_df
        self.max_df = max_df
        self.cache_path = os.path.join(output_dir, 'content_scorer_cache.pkl')

    def fit(self, X, y=None):
        print('start fit ContentBasedScorer', datetime.now().strftime('%H:%M'))
        self.user_history_ = X.groupby('user_id')['book_id'].apply(list).to_dict()

        books_with_tags = self.books_df[['book_id', 'goodreads_book_id', 'title']].merge(
            self.book_tags_df[['goodreads_book_id', 'tag_id', 'count']],
            on='goodreads_book_id',
            how='left'
        ).merge(
            self.tags_df[['tag_id', 'tag_name']],
            on='tag_id',
            how='left'
        )
        profiles = {}
        for _, row in books_with_tags.iterrows():
            book_id = row['book_id']
            if book_id not in profiles:
                title = row['title']
                profiles[book_id] = {
                    'title': title,
                    'weighted_tags': [],
                    'all_tags': set()
                }
            tag_name = str(row['tag_name']).lower().replace(' ', '_')
            count = max(int(row['count']), 1)
            log_weight = int(np.log2(count))
            profiles[book_id]['weighted_tags'].extend([tag_name] * log_weight)
            profiles[book_id]['all_tags'].add(row['tag_name'])
        book_profiles = []
        for book_id, data in profiles.items():
            book_profiles.append({
                'book_id': book_id,
                'title': data['title'],
                'profile': f"{data['title']} {' '.join(data['weighted_tags'])}".strip(),
                'tags_count': len(data['all_tags']),
                'all_tags': ' '.join(data['all_tags'])
            })
        self.book_profiles_df_ = pd.DataFrame(book_profiles).set_index('book_id')

        self.tfidf_vectorizer_ = TfidfVectorizer(
            max_features=self.max_features,
            min_df=self.min_df,
            max_df=self.max_df,
            stop_words='english',
            ngram_range=(1, 3)
        )
        self.tfidf_matrix_ = self.tfidf_vectorizer_.fit_transform(self.book_profiles_df_['profile'])
        self.bookid2idx_ = {bid: i for i, bid in enumerate(self.book_profiles_df_.index)}
        print('end fit ContentBasedScorer', datetime.now().strftime('%H:%M'))
        return self

    def _user_content_scores(self, user_id, group):
      idxs = group.index.values
      scores = np.zeros(len(group))
      user_books = self.user_history_.get(user_id, [])[-20:]
      user_books = [bid for bid in user_books if bid in self.bookid2idx_]
      if len(user_books) == 0:
          return idxs, scores

      user_books_vecs = self.tfidf_matrix_[[self.bookid2idx_[bid] for bid in user_books]]
      user_profile = user_books_vecs.mean(axis=0)
      candidate_idxs = [self.bookid2idx_.get(row['book_id'], None) for _, row in group.iterrows()]
      valid_mask = [i for i, idx in enumerate(candidate_idxs) if idx is not None]
      if not valid_mask:
          return idxs, scores
      candidate_vecs = self.tfidf_matrix_[[candidate_idxs[i] for i in valid_mask]]

      sim = candidate_vecs @ user_profile.T
      sim = np.array(sim).flatten()
      for j, i in enumerate(valid_mask):
          scores[i] = sim[j]

      return idxs, scores

    def transform(self, X):
      print('start transform ContentBasedScorer', datetime.now().strftime('%H:%M'))
      use_cache = getattr(self, 'use_cache', True)
      sim_scores = np.zeros(X.shape[0])
      groups = list(X.groupby('user_id'))

      if use_cache and os.path.exists(self.cache_path):
          with open(self.cache_path, 'rb') as f:
              user_cache = pickle.load(f)
          print(f"Кеш загружен, пользователей в кеше: {len(user_cache)}")
      else:
          user_cache = {}

      to_compute = [(user_id, group) for user_id, group in groups if user_id not in user_cache]
      if use_cache:
        print(f"Пользователей для расчёта: {len(to_compute)} (остальные из кеша)")
      else:
        print(f"Пользователей для расчёта: {len(to_compute)} (кеш отключен)")
      results = [self._user_content_scores(user_id, group) for user_id, group in tqdm(to_compute)]
      for (user_id, group), (idxs, scores) in zip(to_compute, results):
          user_cache[user_id] = scores

      if use_cache:
        with open(self.cache_path, 'wb') as f:
            pickle.dump(user_cache, f)
            print(f"Кеш сохранён, всего пользователей: {len(user_cache)}")

      pos = 0
      for user_id, group in groups:
          n = len(group)
          scores = user_cache[user_id]
          sim_scores[pos:pos+n] = scores
          pos += n

      X['content_score'] = sim_scores
      print('end transform ContentBasedScorer', datetime.now().strftime('%H:%M'))
      return X
```

```python
class TopNPopularityScorer(BaseEstimator, TransformerMixin):
    def __init__(self, min_ratings_threshold=250):
        self.min_ratings_threshold = min_ratings_threshold

    def fit(self, X, y=None):
        print('start fit TopNPopularityScorer', datetime.now().strftime('%H:%M'))
        book_popularity = (
            X.groupby('book_id')
             .agg(
                 rating_count=('rating', 'count'),
                 avg_rating=('rating', 'mean'),
             )
             .reset_index()
        )
        filtered_books = book_popularity[book_popularity['rating_count'] >= self.min_ratings_threshold]
        filtered_books = filtered_books.sort_values('avg_rating', ascending=False)

        self.book2rank_ = {book_id: rank for rank, book_id in enumerate(filtered_books['book_id'])}
        self.max_rank_ = len(self.book2rank_)
        print('end fit TopNPopularityScorer', datetime.now().strftime('%H:%M'))
        return self

    def transform(self, X):
        print('start transform TopNPopularityScorer', datetime.now().strftime('%H:%M'))
        scores = []
        for book_id in X['book_id']:
            rank = self.book2rank_.get(book_id, None)
            if rank is not None:
                score = 1.0 / (rank + 1)
            else:
                score = 0.0
            scores.append(score)

        X['topn_score'] = scores
        print('end transform TopNPopularityScorer', datetime.now().strftime('%H:%M'))
        return X
```

```python
class ItemBasedCFScorer(BaseEstimator, TransformerMixin):
    def __init__(self, k=5):
        self.k = k
        self.cache_path = os.path.join(output_dir, 'cf_scorer_cache.pkl')


    def fit(self, X, y=None):
        print('start fit ItemBasedCFScorer', datetime.now().strftime('%H:%M'))
        self.user_item_matrix_ = X.pivot_table(
            index='user_id',
            columns='book_id',
            values='rating',
            fill_value=0
        )
        data = self.user_item_matrix_.values.T.astype(np.float32)
        user_means = data.mean(axis=1, keepdims=True)
        normalized_data = data - user_means
        sim_matrix = cosine_similarity(normalized_data)
        n_books = sim_matrix.shape[0]
        # используем разряженные матрицы для экономии ресурсов
        sparse_sim = lil_matrix((n_books, n_books), dtype=np.float32)
        for i in range(n_books):
            top_k = np.argsort(sim_matrix[i])[-self.k-1:-1]
            sparse_sim[i, top_k] = sim_matrix[i, top_k]
        self.similarity_matrix_ = sparse_sim.tocsr()
        self.bookid2idx_ = {bid: idx for idx, bid in enumerate(self.user_item_matrix_.columns)}
        self.userid2idx_ = {uid: idx for idx, uid in enumerate(self.user_item_matrix_.index)}
        print('end fit ItemBasedCFScorer', datetime.now().strftime('%H:%M'))
        return self

    def _user_cf_scores(self, user_id, group):
        idxs = group.index.values
        scores = np.zeros(len(group))
        if user_id not in self.userid2idx_:
            return idxs, scores
        user_ratings = self.user_item_matrix_.loc[user_id].values
        for i, (idx, row) in enumerate(group.iterrows()):
            book_id = row['book_id']
            if book_id not in self.bookid2idx_:
                scores[i] = 0.0
                continue
            book_index = self.bookid2idx_[book_id]
            book_similarities = self.similarity_matrix_[book_index].toarray().ravel()
            valid_indices = []
            valid_similarities = []
            for other_book_index, similarity in enumerate(book_similarities):
                rating = user_ratings[other_book_index]
                if similarity > 0 and rating > 0 and other_book_index != book_index:
                    valid_indices.append(other_book_index)
                    valid_similarities.append(similarity)
            if len(valid_similarities) == 0:
                scores[i] = 0.0
                continue
            top_k_indices = np.argsort(valid_similarities)[::-1][:self.k]
            similarity_sum = 0.0
            weighted_sum = 0.0
            for j in top_k_indices:
                other_book_index = valid_indices[j]
                similarity = valid_similarities[j]
                rating = user_ratings[other_book_index]
                similarity_sum += similarity
                weighted_sum += similarity * rating
            if similarity_sum == 0.0:
                scores[i] = 0.0
            else:
                scores[i] = weighted_sum / similarity_sum
        return idxs, scores

    def transform(self, X):
      print('start transform ItemBasedCFScorer', datetime.now().strftime('%H:%M'))
      use_cache = getattr(self, 'use_cache', True)
      sim_scores = np.zeros(X.shape[0])
      groups = list(X.groupby('user_id'))

      if use_cache and os.path.exists(self.cache_path):
          with open(self.cache_path, 'rb') as f:
              user_cache = pickle.load(f)
          print(f"Кеш загружен, пользователей в кеше: {len(user_cache)}")
      else:
          user_cache = {}

      to_compute = [(user_id, group) for user_id, group in groups if user_id not in user_cache]
      if use_cache:
        print(f"Пользователей для расчёта: {len(to_compute)} (остальные из кеша)")
      else:
        print(f"Пользователей для расчёта: {len(to_compute)} (кеш отключен)")
      results = [self._user_cf_scores(user_id, group) for user_id, group in tqdm(to_compute)]

      for (user_id, group), (idxs, scores) in zip(to_compute, results):
          user_cache[user_id] = scores

      if use_cache:
          with open(self.cache_path, 'wb') as f:
              pickle.dump(user_cache, f)
          print(f"Кеш сохранён, всего пользователей: {len(user_cache)}")

      # Собираем все результаты (из кеша)
      pos = 0
      for user_id, group in groups:
          n = len(group)
          scores = user_cache[user_id]
          sim_scores[pos:pos+n] = scores
          pos += n

      X['cf_score'] = sim_scores
      print('end transform ItemBasedCFScorer', datetime.now().strftime('%H:%M'))
      return X
```

```python
class SVDBasedScorer(BaseEstimator, TransformerMixin):
    def __init__(self, n_factors=100, n_epochs=20, lr_all=0.005, reg_all=0.02, random_state=42):
        self.n_factors = n_factors
        self.n_epochs = n_epochs
        self.lr_all = lr_all
        self.reg_all = reg_all
        self.random_state = random_state

    def fit(self, X, y=None):
        print('start fit SVDBasedScorer', datetime.now().strftime('%H:%M'))
        surprise_data = X[['user_id', 'book_id', 'rating']].copy()
        reader = Reader(rating_scale=(1, 5))
        data = Dataset.load_from_df(surprise_data, reader)
        trainset = data.build_full_trainset()
        self.model_ = SVD(
            n_factors=self.n_factors,
            n_epochs=self.n_epochs,
            lr_all=self.lr_all,
            reg_all=self.reg_all,
            random_state=self.random_state
        )
        self.model_.fit(trainset)
        print('end fit SVDBasedScorer', datetime.now().strftime('%H:%M'))
        return self

    def transform(self, X):
        print('start transform SVDBasedScorer', datetime.now().strftime('%H:%M'))
        scores = []
        for idx, row in X.iterrows():
            user_id = row['user_id']
            book_id = row['book_id']
            try:
                score = self.model_.predict(user_id, book_id).est
            except Exception:
                score = 0.0
            scores.append(score)
        X['svd_score'] = scores
        print('end transform SVDBasedScorer', datetime.now().strftime('%H:%M'))
        return X
```

```python
class HybridScorer(BaseEstimator, TransformerMixin):
    def __init__(self, user_type_weights, threshold_new_user, books_df, penalty_same_author=0.5, topn=5):
        self.score_cols = ['topn_score', 'content_score', 'cf_score', 'svd_score']
        self.user_num_ratings_col = 'user_num_ratings'
        self.user_type_weights = user_type_weights
        self.threshold_new_user = threshold_new_user

        self.books_df = books_df.copy().set_index('book_id')
        self.penalty_same_author = penalty_same_author
        self.topn = topn

    def fit(self, X, y=None):
        return self

    def _get_user_type(self, num_ratings):
        return 'new' if num_ratings < self.threshold_new_user else 'active'

    def transform(self, X):
        print('start transform HybridScorer', datetime.now().strftime('%H:%M'))

        X['authors'] = X['book_id'].map(self.books_df['authors'])

        result_rows = []
        for user_id, group in X.groupby('user_id'):
            candidate_indices = set()
            for score_col in self.score_cols:
                topn_idx = group.sort_values(score_col, ascending=False).head(self.topn * 2).index
                candidate_indices.update(topn_idx)
            # Оставляем только кандидатов из объединённого пула
            candidates = group.loc[list(candidate_indices)].copy()

            user_type = self._get_user_type(candidates[self.user_num_ratings_col].iloc[0])
            # используем разные веса для новых и активных пользователей
            weights = self.user_type_weights.get(user_type)
            candidates['hybrid_score'] = candidates[self.score_cols].values @ np.array(weights)

            topn_final = candidates.sort_values('hybrid_score', ascending=False).head(self.topn)
            author_to_indices = {}
            for idx, row in topn_final.iterrows():
                authors = str(row['authors']).split(',')
                authors = [a.strip().lower() for a in authors]
                for author in authors:
                    if author not in author_to_indices:
                        author_to_indices[author] = []
                    author_to_indices[author].append(idx)
            for indices in author_to_indices.values():
                if len(indices) > 1:
                    for i in indices[1:]:
                        # штрафуем за повторяющихся авторов в топ-N
                        candidates.loc[i, 'hybrid_score'] -= self.penalty_same_author * (len(indices) - 1)

            result_rows.append(candidates)

        result = pd.concat(result_rows, ignore_index=True)
        print('end transform HybridScorer', datetime.now().strftime('%H:%M'))
        return result
```

Посчитав результат рекомендаций, самое время посчитать объективные метрики качества.

```python

all_user_recs = {name: {} for name in model_names}

for user_id in valid_user_ids:
    user_result = result[result['user_id'] == user_id]
    for name, col in zip(model_names, model_score_cols):
        top_books = (
            user_result.sort_values(col, ascending=False)
                       .head(top_n)['book_id']
                       .tolist()
        )
        all_user_recs[name][user_id] = top_books

# книги, которые пользователь реально прочитал в тесте
test_truth = test_ratings_df.groupby('user_id')['book_id'].apply(set).to_dict()
```

```python
def precision_at_k(recommended, truth, k):
    if not truth: return np.nan
    return len(set(recommended[:k]) & set(truth)) / k

def recall_at_k(recommended, truth, k):
    if not truth: return np.nan
    return len(set(recommended[:k]) & set(truth)) / len(truth)

def dcg_at_k(recommended, truth, k):
    dcg = 0.0
    for i, rec in enumerate(recommended[:k]):
        if rec in truth:
            dcg += 1.0 / np.log2(i + 2)
    return dcg

def ndcg_at_k(recommended, truth, k):
    ideal_dcg = dcg_at_k(list(truth)[:k], truth, k)
    if ideal_dcg == 0: return np.nan
    return dcg_at_k(recommended, truth, k) / ideal_dcg
```

```python
metrics = {name: {'precision': [], 'recall': [], 'ndcg': []} for name in model_names}

for user_id in valid_user_ids:
    truth = test_truth.get(user_id, set())
    for name in model_names:
        recs = all_user_recs[name][user_id]
        metrics[name]['precision'].append(precision_at_k(recs, truth, top_n))
        metrics[name]['recall'].append(recall_at_k(recs, truth, top_n))
        metrics[name]['ndcg'].append(ndcg_at_k(recs, truth, top_n))
```

1. **Сравнительный анализ метрик**

- **Лучший Precision@5 и Recall@5** показывает модель topn.
- **SVD** также даёт хорошие результаты, уступая только topn.
- **Гибридная модель** (hybrid) не превосходит лучшие отдельные модели и работает примерно на уровне content-based.
- **CF** (item-based) показывает самые низкие значения по всем метрикам.

2. **Анализ эффективности гибридного подхода**

- Гибридная модель не смогла превзойти лучшие отдельные модели (topn, svd).
- Гибрид работает чуть лучше content-based, но хуже topn и svd.
- Это говорит о том, что либо веса в гибриде не оптимальны

3. **Сильные и слабые стороны методов**

**TopN (популярные книги)**
- **Сильные стороны:**  
  - Простота, высокая точность на разреженных данных, всегда рекомендует "безопасные" книги.
- **Слабые стороны:**  
  - Нет персонализации, рекомендации однообразны для всех.

**SVD**
- **Сильные стороны:**  
  - Персонализированные рекомендации, хорошо работает на плотных данных.
- **Слабые стороны:**  
  - Требует много данных, плохо работает для новых пользователей/книг.

**Content-based**
- **Сильные стороны:**  
  - Может рекомендовать новые книги, если есть признаки.
- **Слабые стороны:**  
  - Не учитывает вкусы других пользователей, работает хуже на разреженных данных.

**CF (item-based)**
- **Сильные стороны:**

## Оптимизация весов в гибридной модели

Заново пересчитывать весь пайплайн, затем его применять к тестовой выборке это вечность по времени. Будем использовать легальный, но экономный вариант. У меня уже сохранен результат работы каждого скорера. Я могу подменять веса лишь для гибридной модели и пересчитывать с ними hybrid_score, оптимизируя метрику nDCG@5. Лучший nDCG@5 - лучшие веса. Расчет метрики будет честным. Переобучать ничего не придется и вызывать transform соответственно тоже будет не нужно, достаточно лишь отсорировать книги по новому hybrid_score.

```python
def evaluate_hybrid_weights(result, valid_user_ids, test_truth, score_cols, weights, top_n=5):
    ndcgs = []
    for user_id in valid_user_ids:
        user_result = result[result['user_id'] == user_id].copy()
        # Считаем новый hybrid_score с текущими весами
        user_result['hybrid_score'] = user_result[score_cols].values @ np.array(weights)
        recommended = user_result.sort_values('hybrid_score', ascending=False).head(top_n)['book_id'].tolist()
        truth = test_truth.get(user_id, set())
        ndcgs.append(ndcg_at_k(recommended, truth, top_n))
    return np.nanmean(ndcgs)

n_iter = 250
best_score = -1
best_weights = None
score_cols = ['topn_score', 'content_score', 'cf_score', 'svd_score']

for i in range(n_iter):
    weights = np.random.dirichlet(np.ones(len(score_cols)))
    score = evaluate_hybrid_weights(result, valid_user_ids, test_truth, score_cols, weights, top_n=5)
    print(f"Попытка {i+1}: веса={weights}, nDCG@5={score:.4f}")
    if score > best_score:
        best_score = score
        best_weights = weights

print("\nЛучшие веса:", best_weights)
print("Лучший nDCG@5:", best_score)

result['hybrid_score'] = result[score_cols].values @ best_weights
```

```python
for user_id in valid_user_ids:
    user_result = result[result['user_id'] == user_id]
    top_books = (
        user_result.sort_values('hybrid_score', ascending=False)
                   .head(top_n)['book_id']
                   .tolist()
    )
    all_user_recs['hybrid'][user_id] = top_books

metrics['hybrid'] = {'precision': [], 'recall': [], 'ndcg': []}
for user_id in valid_user_ids:
    truth = test_truth.get(user_id, set())
    recs = all_user_recs['hybrid'][user_id]
    metrics['hybrid']['precision'].append(precision_at_k(recs, truth, top_n))
    metrics['hybrid']['recall'].append(recall_at_k(recs, truth, top_n))
    metrics['hybrid']['ndcg'].append(ndcg_at_k(recs, truth, top_n))

print(f"{'Model':10} | {'Precision@5':>10} | {'Recall@5':>10} | {'nDCG@5':>10}")
print("-"*50)
p = np.nanmean(metrics['hybrid']['precision'])
r = np.nanmean(metrics['hybrid']['recall'])
n = np.nanmean(metrics['hybrid']['ndcg'])
print(f"{'hybrid':10} | {p:10.4f} | {r:10.4f} | {n:10.4f}")
```

## Wide & Deep

```python
model_path = os.path.join(output_dir, 'widedeep_model.pt')
n_epochs = 50
batch_size = 1024
emb_dim = 32
wide_cols = [
    'topn_score',
    'content_score',
    'cf_score',
    'svd_score',
    'user_num_ratings',
    'user_mean_rating',
    'user_activity',
    'book_popularity',
    'book_rating_std'
]
deep_cols = []

test_pairs = set(zip(test_ratings_df['user_id'], test_ratings_df['book_id']))
# Добавляем колонку target: 1 если пара есть в тесте, иначе 0
result['target'] = result.apply(lambda row: (row['user_id'], row['book_id']) in test_pairs, axis=1).astype(int)

# Кодируем user_id и book_id для эмбеддингов
user_encoder = LabelEncoder()
item_encoder = LabelEncoder()
result['user_idx'] = user_encoder.fit_transform(result['user_id'])
result['item_idx'] = item_encoder.fit_transform(result['book_id'])

# Формируем train-датасет для обучения нейросети
X_wide = result[wide_cols].values.astype('float32')
X_deep = result[deep_cols].values.astype('float32') if deep_cols else np.zeros((len(result), 0), dtype='float32')
user_idx = result['user_idx'].values
item_idx = result['item_idx'].values
y = result['target'].values.astype('float32')
```

```python
class WideDeepDataset(Dataset):
    def __init__(self, X_wide, X_deep, user_idx, item_idx, y):
        self.X_wide = torch.tensor(X_wide, dtype=torch.float32)
        self.X_deep = torch.tensor(X_deep, dtype=torch.float32)
        self.user_idx = torch.tensor(user_idx, dtype=torch.long)
        self.item_idx = torch.tensor(item_idx, dtype=torch.long)
        self.y = torch.tensor(y, dtype=torch.float32)
    def __len__(self):
        return len(self.y)
    def __getitem__(self, idx):
        return self.user_idx[idx], self.item_idx[idx], self.X_wide[idx], self.X_deep[idx], self.y[idx]

dataset = WideDeepDataset(X_wide, X_deep, user_idx, item_idx, y)
loader = DataLoader(dataset, batch_size=batch_size, shuffle=True)

class WideAndDeep(nn.Module):
    def __init__(self, n_users, n_items, n_wide, n_deep, emb_dim=32):
        super().__init__()
        self.user_emb = nn.Embedding(n_users, emb_dim)
        self.item_emb = nn.Embedding(n_items, emb_dim)
        self.deep = nn.Sequential(
            nn.Linear(2*emb_dim + n_deep, 128),
            nn.ReLU(),
            nn.Linear(128, 64),
            nn.ReLU(),
            nn.Linear(64, 1)
        )
        self.wide = nn.Linear(n_wide, 1)
    def forward(self, user_idx, item_idx, wide_x, deep_x):
        user_vec = self.user_emb(user_idx)
        item_vec = self.item_emb(item_idx)
        deep_input = torch.cat([user_vec, item_vec, deep_x], dim=1)
        deep_out = self.deep(deep_input)
        wide_out = self.wide(wide_x)
        return deep_out + wide_out
```

```python
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Используем устройство: {device}")

model = WideAndDeep(
    n_users=len(user_encoder.classes_),
    n_items=len(item_encoder.classes_),
    n_wide=X_wide.shape[1],
    n_deep=X_deep.shape[1],
    emb_dim=emb_dim
).to(device)

optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)
loss_fn = nn.BCEWithLogitsLoss()

if os.path.exists(model_path):
    print("Загружаю обученную Wide&Deep модель...")
    model.load_state_dict(torch.load(model_path, map_location=device))
else:
    print("Обучаю Wide&Deep модель...")
    for epoch in range(n_epochs):
        model.train()
        total_loss = 0
        for user_idx_batch, item_idx_batch, X_wide_batch, X_deep_batch, y_batch in tqdm(loader, desc=f"Epoch {epoch+1}/{n_epochs}"):
            user_idx_batch = user_idx_batch.to(device)
            item_idx_batch = item_idx_batch.to(device)
            X_wide_batch = X_wide_batch.to(device)
            X_deep_batch = X_deep_batch.to(device)
            y_batch = y_batch.to(device)
            optimizer.zero_grad()
            logits = model(user_idx_batch, item_idx_batch, X_wide_batch, X_deep_batch).squeeze()
            loss = loss_fn(logits, y_batch)
            loss.backward()
            optimizer.step()
            total_loss += loss.item() * len(y_batch)
        print(f"Epoch {epoch+1}: loss = {total_loss/len(dataset):.4f}")
    torch.save(model.state_dict(), model_path)
    print(f"Модель сохранена в {model_path}")
```

```python
model.eval()
with torch.no_grad():
    user_idx_tensor = torch.tensor(user_idx, dtype=torch.long).to(device)
    item_idx_tensor = torch.tensor(item_idx, dtype=torch.long).to(device)
    X_wide_tensor = torch.tensor(X_wide, dtype=torch.float32).to(device)
    X_deep_tensor = torch.tensor(X_deep, dtype=torch.float32).to(device)
    scores = model(
        user_idx_tensor,
        item_idx_tensor,
        X_wide_tensor,
        X_deep_tensor
    ).cpu().numpy().flatten()
result['widedeep_score'] = scores
```

```python
all_user_recs['widedeep'] = {}
for user_id in valid_user_ids:
    user_result = result[result['user_id'] == user_id]
    top_books = (
        user_result.sort_values('widedeep_score', ascending=False)
                   .head(top_n)['book_id']
                   .tolist()
    )
    all_user_recs['widedeep'][user_id] = top_books

metrics['widedeep'] = {'precision': [], 'recall': [], 'ndcg': []}
for user_id in valid_user_ids:
    truth = test_truth.get(user_id, set())
    recs = all_user_recs['widedeep'][user_id]
    metrics['widedeep']['precision'].append(precision_at_k(recs, truth, top_n))
    metrics['widedeep']['recall'].append(recall_at_k(recs, truth, top_n))
    metrics['widedeep']['ndcg'].append(ndcg_at_k(recs, truth, top_n))

print(f"{'Model':10} | {'Precision@5':>10} | {'Recall@5':>10} | {'nDCG@5':>10}")
print("-"*50)
for name in model_names + ['widedeep']:
    p = np.nanmean(metrics[name]['precision'])
    r = np.nanmean(metrics[name]['recall'])
    n = np.nanmean(metrics[name]['ndcg'])
    print(f"{name:10} | {p:10.4f} | {r:10.4f} | {n:10.4f}")
```
